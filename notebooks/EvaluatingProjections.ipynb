# Evaluating Projections/Planning Scenarios

## Mathematical Preliminaries


### Notation

- $y$ is any outcome value that we are interested in projecting.
- $y^*$ is the actual realized value of some outcome we are projecting.
- $\mathbf{X}$ is some set of covariates or assumptions on which a projection or realization is based.
- $\mathbf{X}^*$ is a realized/observed set of covariates.
- $\mathbf{\cal{X}}$ is the set of all possible states of the universe. 
- $\cal{F}$ is a proper forecast.
- $\cal{P}(\mathbf{X})$ is a projection under set of covariates or assumptions $\mathbf{X}$
- $\cal{p}(\mathbf{X})$ is a point projection under a set of  covariates or assumptions $\mathbf{X}$

### Definition of a projection



A projection, $\cal{P}(\mathbf{X})$ is a statement of the probability of each event within some set or range of events (e.g., number of hospitalizations due to some cause, the value of a stock) over
some set of future times, under some set of assumptions or conditions. That is a projections states:

$$
\cal{P}(\mathbf{X}) = \forall y\in \mathbf{Y} \Pr(y|\mathbf{X})
$$

where $\mathbf{Y}$ is the set of events over which we have projections and $\mathbf{X}$ is the set of assumptions/covariate values under which those projections are meant to apply. 

Under this notation the probability of a single projected event is indicated as:
$$\cal{P}(y | \mathbf{X}) =  \Pr(y|\mathbf{X})$$

We define a single value projection, $\cal{p}(\mathbf{X})$ as a special case where for some value, $y_i$, $\Pr(y_i|\mathbf{X}) = 1$ and
$\forall y_j \text{ where } y_j \neq y_i, \Pr(y_j|\mathbf{X}) = 0$.


### Definition of a forecast

A proper orecast is a projection where the set of conditional values $\mathbf{X}$ includes all possible states of the universe, $\mathbf{\mathcal{X}}$ given the current state of the universe. That is, a forecast givens us:

$$\cal F(y) = \int_{\mathbf{X} \in \mathbf{\mathcal{X}}} \Pr(y|\mathbf{X})\Pr(\mathbf{X})$$

and a full forecast is the set of forecasts for all events:

$$\cal{F} = \forall y \in \mathbf{Y} \Pr(y|\mathbf{X})\Pr(\mathbf{X})$$

### Definition of a scenario projection

A scenario projection is a projection that is contingent on some set of current or future conditions that are unknown at the time at which the projections is made. That is for some set of assumptions, $\mathbf{X}_i$ the scenario projection is:
$$\cal{P(\mathbf{X}_i)}$$

Note that there is always some implicit set of covariates in $\mathbf{\cal{X}}$$ for which we don't make assumptions as part of a scenario projection, but we don't bother with this in the notation. 

## Evaluating Counterfactual Scenarios

We begin with the simplest case, where projections are made along a single scenario axis (i.e., $\mathcal{X}$ can be defined by some set $x = \{x_1, ..., x_n\}$). First consider a set of two scenario projections, $\mathcal{p}(x_1)$ and $\mathcal{p}(x_2)$. To evaluate these point projections, we consider the observed value $y^* = y(x^*)$, which has a corresponding scenario value $x^*$ that is known. We also presume there are true values at each of the scenarios, $y(x_1)$ and $y(x_2)$; these true values are unknown. We can define *miscalibration bias*, $m$, which is the difference between the projected value at the observed scenario and the observed value, $\mathcal{p}(x^*) - y^*$. 

Ultimately, we are interested in the true bias of each projection, $\epsilon_i = \mathcal{p}(x_i) - y(x_i)$, but this cannot be measured directly because $y(x_i)$ is unknown. We can, however, draw conclusions about these errors under certain assumptions.

First, we assume that both $\mathcal{p}$ and $y$ are linear. For simplicity, we define the linear function relative to the first scenario, i.e., $\mathcal{p}(x) = \mathcal{p}(x_1) + \frac{\mathcal{p}(x_2)-\mathcal{p}(x_1)}{x_2-x_1}(x)$. Then, we can show that miscalibration bias is a weighted average of the true bias for each scenario, 

$$
\begin{align}
m &= \mathcal{p}(x^*) - y^* \\
  &= \mathcal{p}(x_1) + \frac{\mathcal{p}(x_2)-\mathcal{p}(x_1)}{x_2-x_1}\Big(x^*-x_1\Big) -  y(x_1) + \frac{y(x_2)-t(x_1)}{x_2-x_1}\Big(x^*-x_1\Big) \\
  &= \Big(1-\frac{x^*-x_1}{x_2-x_1}\Big)\Big(\mathcal{p}(x_1) - y(x_1)\Big) + \Big(\frac{x^*-x_1}{x_2-x_1}\Big)\Big(\mathcal{p}(x_2) - y(x_2)\Big) \\
  &= \Big(1-\frac{x^*-x_1}{x_2-x_1}\Big) \epsilon_1 + \Big(\frac{x^*-x_1}{x_2-x_1}\Big)\epsilon_2 
\end{align}
$$

However, there are infinitely many combinations of $\mathcal{p}(x_1)$ and $\mathcal{p}(x_2)$ that satisfy this equation. In other words, small miscalibration bias does not ensure small true bias, and vice versa. By assuming $\mathcal{p}(x) > 0$ for all $x$, we can constrain the possible outcomes.

#### Try/Thoughts

Consider a situation where we are unable to modify the probability of the outcomes, and our ability to optimize utility is
completely based on the results of designing around some future value $y$. The realized utility of designing around a 
value $y_d$ is captured by some function $U(y_d, y_a)$ where $y_a$ is the actual value. 

For example, $y$ might be the expected number of people presenting for hospitalization. There is some cost of maintaining an 
extra hospital bed, say 5,000, and an expected cost of someone presenting for hospitalization if we do not have a bed available
reflecting emergency allocation of resources and the increased liklihood of server oucomes, say 25,000 per excess case. Hence $U(y_d, y_a)$
would be:
$$ 
U(y_d, y_a) = 
\begin{cases}
    (-5000)(y_d-y_a) &\text{if } y_d \geq y_a\\
    (-25000)(y_a-y_d) &\text{if } y_a > y_d
\end{cases}
$$

Based on this utility function, the expected utility of some decision given a probabilistic prediction $\mathscr{P}$ is:

$$
E(U(y_d)) = \sum_{y \in Y} \Pr(y|\mathscr{P})U(y_d,y)
$$

Where $Y$ is the set of all possible outcomes and $\Pr(y|\mathscr{P})$
is the probability of $y$ if the prediction $\mathscr{P}$ is correct.

Note that this implies an optimal value to design around, $y^*_d$:
$$
y^*_d|\mathscr{P} = \argmax_{y_d}  \sum_{y \in Y} \Pr(y|\mathscr{P})U(y_d,y)
$$

Given this, the error in predicted utility of any probabilistic
prediction can be considered to be:

$$
\left| U(y^*_d|\mathscr{P}, y_a) -  \sum_{y \in Y} \Pr(y|\mathscr{P})U(y^*_d|\mathscr{P},y) \right|
$$


Some properties of this approach:

- The projection implied optimal value to base decisions on is the weighted mean, which in the case of a simple linear error function should be the mean forecast (not the median).

- The error function will penalize under certainty when 
the forecast is correct.
