## Preparing versus Intervening

$$
U = \sum_{y \in Y} u(y) \Pr(y) = \sum_{y \in Y} \sum_{\mathbf{X} \in \mathbf{\cal{X}}} \left[ c(\mathbf{X}) + u(y|\mathbf{X}) \Pr(y|\mathbf{X})  \right] \Pr(\mathbf{X})
$$

- $U$ is the expected utility at some future point in time
- $Y$ is some random variable we are interested in projecting
- $\mathbf{X}$ defines some set of covariates or state of the world under which $y$ may occur
- $\mathbf{\cal{X}}$ represents the space of possible states of the world $\mathbf{X}$ can take
- $u(y|\mathbf{X})$ is the utility of having outcome $y$ given state of the world $\mathbf{X}$
- $c(\mathbf{X})$ is the cost associated with achieving and being in the state of the world $\mathbf{X}$, including both extrinsic costs with being in that state of the world (e.g., side effects of schools being closed) and costs associated with achieving $\mathbf{X}$.

**Equation 1:** Equation for overall utility at some future state of the world.

If we operate in the framework that the goal of our decisions is the optimize future utility, these goals of these decisions can be divided into two categories, "preparing" and "intervening". *Preparing* encompasses decisions that attempt to optimize some entity's utility under some set of future events (i.e., optimize $u(Y)$), such as bringing an umbrella to work when it may rain, or an individual making an investment in their retirement account.  *Intervening*  encompasses decisions that attempt to improve utility by influencing the future conditions that impact some outcome (i.e., optimize $\Pr(Y)$), such as mandating social distancing or reducing carbon emissions. It is worth noting that many actions have both the function of preparing and intervening. For instance, vaccine both decreases the probability of infection and the health impact of that infection should it occur.

To be more precise in our treatment of how preparers and interveners act on the world...[DEFINE ACTIONS MORE EXPLICITLY IN THE EQUATIONS]

$$
U_\textbf{a} = \sum_{y \in Y} u(y) \Pr(y) = \sum_{y \in Y} \sum_{\mathbf{X} \in \mathbf{\cal{X}}} \left[ c(\mathbf{X},\textbf{a}) + u(y|\mathbf{X}) \Pr(y|\mathbf{X})  \right] \Pr(\mathbf{X}|\textbf{a})
$$

**Equation 2:** Extension of Eq. 1 to explicitly show how an action of set of actions, $\mathbf{a}$ influence the expected utility.

## Assessing the Value of Scenario Projections

We are interested in how much scenario models contribute to our correct assessment of $U_1 - U_2$ where $U_1$ and $U_2$ are the expected utility under actions $a_1$ and $a_2$.

We will first focus on the case where we are concerned with comparing the utility when the target conditions of action(s) $\mathbf{a}$, $\mathbf{X_a}$ are deterministically achievable. We will further assume that $u(y)$ is independent of $\mathbf{X_a}$ and $\mathbf{a}$ beyond the cost function $c(\mathbf{X}_a, \mathbf{a})$, which is independent of $y$.

Under this assumption:

$$
\begin{align*}
U_\textbf{a} & =  \sum_{y \in Y} \sum_{\mathbf{X} \in \mathbf{\cal{X}}} \left[ c(\mathbf{X},\textbf{a}) + u(y|\mathbf{X}) \Pr(y|\mathbf{X})  \right] \Pr(\mathbf{X}|\textbf{a}) \\
& = c(\mathbf{X}_a, \mathbf{a}) + \sum_{y \in \mathbf{Y}} u(y)\Pr(y|\mathbf{X}_a)
\end{align*}
$$


Under these assumptions we now want to consed the value of our models in correctly estimating the utility of some decsiion between actions. Let us define:

- $U_a$ to be the prior expected utility before we have a model.
- $U^\prime_a $ to be the expected utility after we have a scenario model of conditions $\mathbf{X_a}$ (is this the pre-posterior???).
- $U^*_a$ to be the expected uitlity of action $\textbf{a}$ after we have seen the actual state of the world $\textbf{X}^*$ and $y^*$.

We postualte that a valid estimate of the value of having models was by characterizing how much it moved our estimate of the utility difference between two decisions toward the expected utility after observing states of the world. That is by comparing:

$$
|(U_1 - U_2) - (U^*_1 - U^*_2)|
$$

versus

$$
|(U_1^\prime - U_2^\prime) - (U^*_1 - U^*_2)|
$$

Under the above assumptions, we have:
$$
\begin{align*}
(U_1 - U_2) &= (c(\mathbf{X}_1, \mathbf{a}_1)- c(\mathbf{X}_2, \mathbf{a}_2)) + \sum_{y\in \mathbf{Y}} \left[  u(y) \Pr(y|\mathbf{X}_1) - u(y) \Pr(y|\mathbf{X}_2) \right]\\
&= \Delta c(\mathbf{a}_1,\mathbf{a}_2) + \sum_{y\in \mathbf{Y}} u(y) \left[  \Pr(y|\mathbf{X}_1) - \Pr(y|\mathbf{X}_2) \right]
\end{align*}
$$

where $\Delta c(\mathbf{a}_1,\mathbf{a}_2) = (c(\mathbf{X}_1, \mathbf{a}_1)- c(\mathbf{X}_2, \mathbf{a}_2))$. Hence, if we presume that observing the world does not modify our estimate of the cost of an action and its resulting state of the world:

$$
\begin{align*}
(U_1-U_2) - (U^*_1 - U^*_2) &= \Delta c(\mathbf{a}_1,\mathbf{a}_2) + \sum_{y\in \mathbf{Y}} u(y) \left[  \Pr(y|\mathbf{X}_1) - \Pr(y|\mathbf{X}_2) \right] - \Delta c(\mathbf{a}_1,\mathbf{a}_2) + \sum_{y\in \mathbf{Y}} u(y) \left[  \Pr(y|\mathbf{X}_1, \mathbf{X}^*) - \Pr(y|\mathbf{X}_2, \mathbf{X}^*) \right] \\
&= \sum_{y\in \mathbf{Y}} u(y) \left[ \Pr(y|\mathbf{X}_1) -  \Pr(y|\mathbf{X}_1, \mathbf{X}^*) +  \Pr(y|\mathbf{X}_2, \mathbf{X}^*) -  \Pr(y|\mathbf{X}_2)  \right]
\end{align*}
$$

### Evaluating for preparers (a brief digression)

#### Proof of the obvious: in the simple case preparers only care about forecasts

Let's consider that in the preparers universe states of the world $\mathbf{X}$ can be partitioned into $<\mathbf{X}_A, \mathbf{X}_B>$ where $Y \perp \mathbf{X}_A$ and any action we might take is assumed only to impact $\mathbf{X}_A$ (i.e., $\mathbf{X}_B \perp \mathbf{a}$). Under these assumptions:

$$
\begin{align*}
U_\textbf{a} &= \sum_{y \in Y} u(y) \Pr(y) = \sum_{y \in Y} \sum_{\mathbf{X} \in \mathbf{\cal{X}}} \left[ c(\mathbf{X},\textbf{a}) + u(y|\mathbf{X}) \Pr(y|\mathbf{X}) \right]  \Pr(\mathbf{X}|\textbf{a}) \\
 &= \sum_{y \in Y} \sum_{\mathbf{X_A} \in \mathbf{\cal{X_A}}}  \sum_{\mathbf{X_B} \in \mathbf{\cal{X_B}}} \left[ c(\mathbf{X_A},\mathbf{X_B},\textbf{a}) + u(y|\mathbf{X_A}, \mathbf{X_B}) \Pr(y|\mathbf{X_A}, \mathbf{X_B} )  \right] \Pr(\mathbf{X_A}, \mathbf{X_B}|\textbf{a})\\
 &= \sum_{y \in Y} \sum_{\mathbf{X_A} \in \mathbf{\cal{X_B}}}  \sum_{\mathbf{X_B} \in \mathbf{\cal{X_B}}} \left[ c(\mathbf{X_A},\mathbf{X_B},\textbf{a}) + u(y|\mathbf{X_A}, \mathbf{X_B}) \Pr(y|\mathbf{X_B} ) \right]\Pr(\mathbf{X_B}) \Pr(\mathbf{X_A}|\textbf{a}) \\
&= \sum_{y \in Y} \sum_{\mathbf{X_A} \in \mathbf{\cal{X_A}}} \Pr(\mathbf{X_A}|\textbf{a})  \left[E(c(\mathbf{X_A},\textbf{a})) + E(u(y|\textbf{X}_A))\Pr(y)\right]
\end{align*}
$$

Presuming $\textbf{X}_A$ to be deterministically obtained given $\textbf{a}$ and noting that $c()$ and $u()$ already denote expectations:
$$
U_\textbf{a} = c(\textbf{X}_A, \textbf{a}) \sum_{y \in Y}  u(y|\textbf{X}_A)\Pr(y)
$$

## Conclusion and Discussion

- Most of the ways in which sets scenario projections can be "wrong", in the sense of providing non-informative or misleading information towards a decision, have nothing to do with the accuracy/calibration of the scenarios themselves.
- Still, even if conditional model calibration is not *sufficient* for having an informative set of scenarios, it is a *necessary* component.
    - not all errors matter, but only by being fully accurate would the scenarios provide valid/useful information to people with all utility functions (i.e., the argument that a set of scenarios can lead to the right decision but be wrong is only correct under a subset of utility/cost functions).
- Many of the ways that a set of scenarios can go wrong have to do with the definition of the scenarios themselves.
