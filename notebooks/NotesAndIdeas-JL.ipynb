## Perspective of Decision Makers

- VOI analyses tend to be highly based on the idea that there is a known and fixed decision point to plan around. However, this is not necessarily how information in passed up the chain to high level decision makers. Often they are given a set of information of the flavor "you will save y lives if you do X" and then use their own judgement. Hence, classic decision point based analysis may be less relevant than usually argued.
