## "Dangerous" scenarios for model use
- You are operating under the assumption that you can only modify $u(y)$ but in fact you are also modifying $\Pr(\mathbf{X})$
- You make substantial investment (i.e., spend substantial $c(X)$) in optimizing $u(y)$, but others are modifying $\Pr(\mathbf{X})$ based on an assessment of $\Pr(y)$.
